```python


# ============================================================================
# SECTION 1: SETUP AND INSTALLATIONS
# ============================================================================

# Install required packages
import subprocess
import sys

def install_package(package):
    subprocess.check_call([sys.executable, "-m", "pip", "install", package])

# Install only necessary packages
required_packages = [
    "torch", "torchvision", "torchaudio",
    "opencv-python-headless", "albumentations",
    "scikit-learn", "matplotlib", "seaborn", "tqdm"
]

for package in required_packages:
    try:
        __import__(package.replace("-", "_"))
    except ImportError:
        print(f"Installing {package}...")
        install_package(package)

# Import libraries
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import torchvision.transforms as transforms
from torchvision.models import resnet50

# Standard libraries
import numpy as np
import cv2
import random
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import roc_auc_score, precision_recall_curve, auc, f1_score
import albumentations as A
from albumentations.pytorch import ToTensorV2
from tqdm import tqdm
import warnings
warnings.filterwarnings('ignore')

print("✅ Libraries imported successfully!")

# Set random seeds for reproducibility
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

set_seed(42)

# Device configuration
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"🔧 Using device: {device}")
if torch.cuda.is_available():
    print(f"🎮 GPU: {torch.cuda.get_device_name()}")

# ============================================================================
# SECTION 2: ENHANCED DATASET CREATION
# ============================================================================

def create_enhanced_texture(size=(256, 256), texture_type='cloth'):
    """Create more realistic texture data"""
    if texture_type == 'cloth':
        # More realistic cloth texture with fibers
        base = np.random.uniform(80, 180, size + (3,)).astype(np.float32)

        # Add fiber patterns
        x, y = np.meshgrid(np.arange(size[1]), np.arange(size[0]))
        fiber_pattern1 = np.sin(x/3) * np.cos(y/4) * 15
        fiber_pattern2 = np.sin((x+y)/5) * 10

        texture = base + fiber_pattern1[:,:,np.newaxis] + fiber_pattern2[:,:,np.newaxis]

        # Add realistic noise
        noise = np.random.normal(0, 8, size + (3,)).astype(np.float32)
        texture = np.clip(texture + noise, 0, 255).astype(np.uint8)

    elif texture_type == 'wafer':
        # More realistic wafer with grid patterns
        base = np.full(size + (3,), 140, dtype=np.float32)

        # Add grid pattern
        x, y = np.meshgrid(np.arange(size[1]), np.arange(size[0]))
        grid_x = np.sin(x * 2 * np.pi / 32) * 5
        grid_y = np.sin(y * 2 * np.pi / 32) * 5

        texture = base + (grid_x + grid_y)[:,:,np.newaxis]
        noise = np.random.normal(0, 3, size + (3,)).astype(np.float32)
        texture = np.clip(texture + noise, 0, 255).astype(np.uint8)

    else:  # metal
        # More realistic metallic texture with brushed patterns
        base = np.random.uniform(60, 140, size + (3,)).astype(np.float32)

        # Add brushed metal pattern
        x, y = np.meshgrid(np.arange(size[1]), np.arange(size[0]))
        brush_pattern = np.sin(x/2) * 20 + np.random.uniform(-5, 5, size)

        texture = base + brush_pattern[:,:,np.newaxis]
        texture = np.clip(texture, 0, 255).astype(np.uint8)

    return texture

def add_realistic_anomaly(image, anomaly_type='scratch'):
    """Add more realistic anomalies"""
    anomaly_img = image.copy()
    mask = np.zeros(image.shape[:2], dtype=np.uint8)

    if anomaly_type == 'scratch':
        # Multiple interconnected scratches
        num_scratches = np.random.randint(1, 4)
        for _ in range(num_scratches):
            start_point = (np.random.randint(30, image.shape[1]-30),
                          np.random.randint(30, image.shape[0]-30))

            # Create curved scratch
            points = [start_point]
            for i in range(np.random.randint(3, 8)):
                last_point = points[-1]
                next_point = (
                    last_point[0] + np.random.randint(-50, 50),
                    last_point[1] + np.random.randint(-50, 50)
                )
                next_point = (
                    max(10, min(image.shape[1]-10, next_point[0])),
                    max(10, min(image.shape[0]-10, next_point[1]))
                )
                points.append(next_point)

            # Draw scratch with varying thickness
            for i in range(len(points)-1):
                thickness = np.random.randint(2, 6)
                cv2.line(anomaly_img, points[i], points[i+1],
                        (np.random.randint(0, 50), np.random.randint(0, 50), np.random.randint(0, 50)),
                        thickness)
                cv2.line(mask, points[i], points[i+1], 255, thickness+2)

    elif anomaly_type == 'hole':
        # Irregular holes with jagged edges
        num_holes = np.random.randint(1, 3)
        for _ in range(num_holes):
            center = (np.random.randint(40, image.shape[1]-40),
                     np.random.randint(40, image.shape[0]-40))

            # Create irregular hole
            angles = np.linspace(0, 2*np.pi, np.random.randint(6, 12))
            radii = np.random.uniform(8, 25, len(angles))

            points = []
            for angle, radius in zip(angles, radii):
                x = int(center[0] + radius * np.cos(angle))
                y = int(center[1] + radius * np.sin(angle))
                points.append([x, y])

            points = np.array(points, dtype=np.int32)
            cv2.fillPoly(anomaly_img, [points], (0, 0, 0))
            cv2.fillPoly(mask, [points], 255)

    elif anomaly_type == 'wrinkle':
        # Complex wrinkle patterns
        num_wrinkles = np.random.randint(2, 5)
        for _ in range(num_wrinkles):
            # Create wrinkle path
            start = (np.random.randint(20, image.shape[1]-20),
                    np.random.randint(20, image.shape[0]-20))

            points = [start]
            for i in range(np.random.randint(5, 10)):
                last_point = points[-1]
                # Make wrinkles more wavy
                angle = np.random.uniform(0, 2*np.pi)
                length = np.random.uniform(10, 30)
                next_point = (
                    int(last_point[0] + length * np.cos(angle)),
                    int(last_point[1] + length * np.sin(angle))
                )
                next_point = (
                    max(10, min(image.shape[1]-10, next_point[0])),
                    max(10, min(image.shape[0]-10, next_point[1]))
                )
                points.append(next_point)

            # Draw wrinkle with shading effect
            points = np.array(points, dtype=np.int32)

            # Main wrinkle line
            cv2.polylines(anomaly_img, [points], False, (40, 40, 40), 3)
            cv2.polylines(mask, [points], False, 255, 5)

            # Add shading around wrinkle
            cv2.polylines(anomaly_img, [points], False, (80, 80, 80), 1)

    elif anomaly_type == 'stain':
        # Add stain anomalies
        center = (np.random.randint(50, image.shape[1]-50),
                 np.random.randint(50, image.shape[0]-50))

        # Create organic stain shape
        angles = np.linspace(0, 2*np.pi, 20)
        radii = np.random.uniform(15, 35, len(angles)) + np.sin(angles*3) * 10

        points = []
        for angle, radius in zip(angles, radii):
            x = int(center[0] + radius * np.cos(angle))
            y = int(center[1] + radius * np.sin(angle))
            points.append([x, y])

        points = np.array(points, dtype=np.int32)

        # Different stain colors
        stain_colors = [(20, 20, 20), (0, 30, 60), (60, 30, 0), (30, 0, 30)]
        color = stain_colors[np.random.randint(0, len(stain_colors))]

        cv2.fillPoly(anomaly_img, [points], color)
        cv2.fillPoly(mask, [points], 255)

    return anomaly_img, mask

def get_enhanced_transforms(img_size=256):
    """Enhanced augmentation transforms"""

    train_transform = A.Compose([
        A.Resize(img_size, img_size),
        A.OneOf([
            A.HorizontalFlip(p=1.0),
            A.VerticalFlip(p=1.0),
            A.RandomRotate90(p=1.0),
            A.Transpose(p=1.0),
        ], p=0.7),

        # Advanced geometric transforms
        A.OneOf([
            A.ElasticTransform(p=1.0, alpha=50, sigma=5),
            A.GridDistortion(p=1.0, num_steps=3, distort_limit=0.1),
            A.OpticalDistortion(p=1.0, distort_limit=0.1),
        ], p=0.3),

        # Advanced color transforms
        A.OneOf([
            A.RandomBrightnessContrast(brightness_limit=0.2, contrast_limit=0.2, p=1.0),
            A.HueSaturationValue(hue_shift_limit=15, sat_shift_limit=25, val_shift_limit=15, p=1.0),
            A.CLAHE(clip_limit=2.0, p=1.0),
        ], p=0.5),

        # Noise and blur
        A.OneOf([
            A.MotionBlur(blur_limit=5, p=1.0),
            A.MedianBlur(blur_limit=3, p=1.0),
            A.GaussianBlur(blur_limit=3, p=1.0),
            A.GaussNoise(var_limit=(10, 30), p=1.0),
        ], p=0.3),

        # Cutout for robustness
        A.CoarseDropout(max_holes=3, max_height=32, max_width=32, p=0.2),

        A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
        ToTensorV2(),
    ])

    val_transform = A.Compose([
        A.Resize(img_size, img_size),
        A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
        ToTensorV2(),
    ])

    return train_transform, val_transform

def create_enhanced_dataset():
    """Create enhanced synthetic dataset with more variety"""
    print("📊 Creating enhanced Texture-AD dataset...")

    normal_images = []
    anomaly_images = []

    texture_types = ['cloth', 'wafer', 'metal']
    anomaly_types = ['scratch', 'hole', 'wrinkle', 'stain']

    # Create more normal samples with variation
    print("Creating normal samples...")
    for i in tqdm(range(2000)):  # Increased from 1000
        texture_type = texture_types[i % len(texture_types)]
        img = create_enhanced_texture(texture_type=texture_type)
        normal_images.append(img)

    # Create more anomaly samples
    print("Creating anomaly samples...")
    for i in tqdm(range(800)):  # Increased from 300
        texture_type = texture_types[i % len(texture_types)]
        anomaly_type = anomaly_types[i % len(anomaly_types)]
        base_img = create_enhanced_texture(texture_type=texture_type)
        anomaly_img, _ = add_realistic_anomaly(base_img, anomaly_type)
        anomaly_images.append(anomaly_img)

    # Split data
    train_images = normal_images[:1600]  # Only normal for training
    train_labels = [0] * 1600

    test_images = normal_images[1600:] + anomaly_images
    test_labels = [0] * 400 + [1] * 800

    print(f"✅ Enhanced dataset created: {len(train_images)} train, {len(test_images)} test")
    return (train_images, train_labels), (test_images, test_labels)

# ============================================================================
# SECTION 3: CUSTOM GRAPH NEURAL NETWORK (Pure PyTorch)
# ============================================================================

class CustomGraphConv(nn.Module):
    """Custom Graph Convolution Layer (Pure PyTorch)"""

    def __init__(self, in_features, out_features, bias=True):
        super(CustomGraphConv, self).__init__()
        self.in_features = in_features
        self.out_features = out_features

        # Linear transformation for node features
        self.linear = nn.Linear(in_features, out_features, bias=bias)

        # Attention mechanism for edge weights
        self.attention = nn.Sequential(
            nn.Linear(in_features * 2, out_features),
            nn.LeakyReLU(0.2),
            nn.Linear(out_features, 1),
            nn.Sigmoid()
        )

        self.reset_parameters()

    def reset_parameters(self):
        nn.init.xavier_uniform_(self.linear.weight)
        if self.linear.bias is not None:
            nn.init.zeros_(self.linear.bias)

    def forward(self, x, edge_index, edge_weight=None):
        """
        x: [num_nodes, in_features]
        edge_index: [2, num_edges]
        edge_weight: [num_edges] (optional)
        """
        # Linear transformation
        h = self.linear(x)

        if edge_index.size(1) == 0:
            return h

        # Message passing
        row, col = edge_index

        # Compute attention weights if not provided
        if edge_weight is None:
            edge_features = torch.cat([x[row], x[col]], dim=1)
            edge_weight = self.attention(edge_features).squeeze()

        # Aggregate messages
        messages = h[col] * edge_weight.unsqueeze(1)

        # Sum aggregation
        out = torch.zeros_like(h)
        out.index_add_(0, row, messages)

        return out + h  # Residual connection

class CustomGraphAttention(nn.Module):
    """Custom Graph Attention Layer (Pure PyTorch)"""

    def __init__(self, in_features, out_features, heads=4, dropout=0.1):
        super(CustomGraphAttention, self).__init__()
        self.in_features = in_features
        self.out_features = out_features
        self.heads = heads
        self.head_dim = out_features // heads

        assert out_features % heads == 0

        self.W = nn.Linear(in_features, out_features, bias=False)
        self.a = nn.Parameter(torch.zeros(heads, 2 * self.head_dim))
        self.dropout = nn.Dropout(dropout)
        self.leakyrelu = nn.LeakyReLU(0.2)

        self.reset_parameters()

    def reset_parameters(self):
        nn.init.xavier_uniform_(self.W.weight)
        nn.init.xavier_uniform_(self.a)

    def forward(self, x, edge_index):
        """
        x: [num_nodes, in_features]
        edge_index: [2, num_edges]
        """
        N = x.size(0)
        h = self.W(x).view(N, self.heads, self.head_dim)

        if edge_index.size(1) == 0:
            return h.view(N, -1)

        row, col = edge_index

        # Compute attention scores
        h_row = h[row]  # [num_edges, heads, head_dim]
        h_col = h[col]  # [num_edges, heads, head_dim]

        # Concatenate and compute attention
        h_cat = torch.cat([h_row, h_col], dim=2)  # [num_edges, heads, 2*head_dim]

        # Attention scores
        e = (h_cat * self.a).sum(dim=2)  # [num_edges, heads]
        e = self.leakyrelu(e)

        # Softmax attention weights per head and per node
        attention = torch.zeros(N, self.heads, device=x.device)
        attention.index_add_(0, row, torch.exp(e))
        attention = attention[row]  # [num_edges, heads]

        # Normalize
        attention = torch.exp(e) / (attention + 1e-16)
        attention = self.dropout(attention)

        # Apply attention
        h_out = torch.zeros(N, self.heads, self.head_dim, device=x.device)
        weighted_h = h_col * attention.unsqueeze(2)
        h_out.index_add_(0, row, weighted_h)

        return h_out.view(N, -1)

class GraphBatch:
    """Custom batch for graph data (Pure PyTorch)"""

    def __init__(self, x_list, edge_index_list, edge_weight_list=None):
        self.batch_size = len(x_list)

        # Concatenate node features
        self.x = torch.cat(x_list, dim=0)

        # Adjust edge indices for batching
        node_offset = 0
        edge_indices = []
        edge_weights = []
        batch_indices = []

        for i, (x, edge_index) in enumerate(zip(x_list, edge_index_list)):
            num_nodes = x.size(0)

            # Adjust edge indices
            if edge_index.size(1) > 0:
                adjusted_edge_index = edge_index + node_offset
                edge_indices.append(adjusted_edge_index)

                # Handle edge weights
                if edge_weight_list is not None and i < len(edge_weight_list):
                    edge_weights.append(edge_weight_list[i])

            # Create batch indices
            batch_indices.extend([i] * num_nodes)
            node_offset += num_nodes

        # Concatenate edge indices
        if edge_indices:
            self.edge_index = torch.cat(edge_indices, dim=1)
            if edge_weights:
                self.edge_weight = torch.cat(edge_weights, dim=0)
            else:
                self.edge_weight = None
        else:
            self.edge_index = torch.zeros(2, 0, dtype=torch.long, device=self.x.device)
            self.edge_weight = None

        self.batch = torch.tensor(batch_indices, device=self.x.device)

def global_mean_pool_custom(x, batch):
    """Custom global mean pooling (Pure PyTorch)"""
    batch_size = batch.max().item() + 1
    out = torch.zeros(batch_size, x.size(1), device=x.device)

    for i in range(batch_size):
        mask = batch == i
        if mask.any():
            out[i] = x[mask].mean(dim=0)

    return out

# ============================================================================
# SECTION 4: ATTENTION MODULES
# ============================================================================

class AttentionModule(nn.Module):
    """Channel and spatial attention module"""

    def __init__(self, channels):
        super(AttentionModule, self).__init__()

        # Channel attention
        self.channel_attention = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Conv2d(channels, channels // 8, 1),
            nn.ReLU(),
            nn.Conv2d(channels // 8, channels, 1),
            nn.Sigmoid()
        )

        # Spatial attention
        self.spatial_attention = nn.Sequential(
            nn.Conv2d(2, 1, 7, padding=3),
            nn.Sigmoid()
        )

    def forward(self, x):
        # Channel attention
        ca = self.channel_attention(x)
        x = x * ca

        # Spatial attention
        avg_pool = torch.mean(x, dim=1, keepdim=True)
        max_pool, _ = torch.max(x, dim=1, keepdim=True)
        spatial_input = torch.cat([avg_pool, max_pool], dim=1)
        sa = self.spatial_attention(spatial_input)
        x = x * sa

        return x

# ============================================================================
# SECTION 5: ENHANCED FPN WITH ATTENTION
# ============================================================================

class EnhancedFPN(nn.Module):
    """Enhanced Feature Pyramid Network with attention"""

    def __init__(self, backbone='resnet50'):
        super(EnhancedFPN, self).__init__()

        if backbone == 'resnet50':
            resnet = resnet50(pretrained=True)
            self.layer0 = nn.Sequential(
                resnet.conv1, resnet.bn1, resnet.relu, resnet.maxpool
            )
            self.layer1 = resnet.layer1  # 256 channels
            self.layer2 = resnet.layer2  # 512 channels
            self.layer3 = resnet.layer3  # 1024 channels
            self.layer4 = resnet.layer4  # 2048 channels

            channels = [256, 512, 1024, 2048]

        # FPN layers with attention
        self.fpn_inner = nn.ModuleList([
            nn.Sequential(
                nn.Conv2d(channels[3-i], 256, 1),
                AttentionModule(256)
            ) for i in range(4)
        ])

        self.fpn_out = nn.ModuleList([
            nn.Sequential(
                nn.Conv2d(256, 256, 3, padding=1),
                nn.BatchNorm2d(256),
                nn.ReLU()
            ) for _ in range(4)
        ])

        # Additional layers
        self.p6 = nn.Conv2d(channels[-1], 256, 3, stride=2, padding=1)
        self.p7 = nn.Conv2d(256, 256, 3, stride=2, padding=1)

    def forward(self, x):
        # Bottom-up pathway
        c0 = self.layer0(x)
        c1 = self.layer1(c0)
        c2 = self.layer2(c1)
        c3 = self.layer3(c2)
        c4 = self.layer4(c3)

        # Top-down pathway with attention
        features = [c1, c2, c3, c4]
        fpn_features = []

        # Start from top
        last_inner = self.fpn_inner[0](c4)
        fpn_features.append(self.fpn_out[0](last_inner))

        for i, (feature, inner_conv, out_conv) in enumerate(
            zip([c3, c2, c1], self.fpn_inner[1:], self.fpn_out[1:])
        ):
            inner_lateral = inner_conv(feature)
            inner_top_down = F.interpolate(
                last_inner, size=inner_lateral.shape[-2:], mode='nearest'
            )
            last_inner = inner_lateral + inner_top_down
            fpn_features.append(out_conv(last_inner))

        # Reverse to get P2, P3, P4, P5
        fpn_features = fpn_features[::-1]

        # Add P6 and P7
        p6 = self.p6(c4)
        p7 = F.relu(self.p7(p6))
        fpn_features.extend([p6, p7])

        return fpn_features

# ============================================================================
# SECTION 6: ENHANCED GRAPH COMPONENTS
# ============================================================================

class EnhancedGraphConstructor(nn.Module):
    """Enhanced graph constructor with better connectivity"""

    def __init__(self, feature_dim=256, node_dim=128):
        super(EnhancedGraphConstructor, self).__init__()
        self.node_dim = node_dim

        self.node_proj = nn.Sequential(
            nn.Conv2d(feature_dim, node_dim, 1),
            nn.BatchNorm2d(node_dim),
            nn.ReLU(),
            nn.Conv2d(node_dim, node_dim, 1),
            nn.BatchNorm2d(node_dim)
        )

        # For edge weight computation
        self.edge_proj = nn.Sequential(
            nn.Linear(node_dim * 2, node_dim),
            nn.ReLU(),
            nn.Linear(node_dim, 1),
            nn.Sigmoid()
        )

    def forward(self, feature_map):
        B, C, H, W = feature_map.shape

        # Extract node features
        node_features = self.node_proj(feature_map)
        node_features = node_features.view(B, self.node_dim, -1).permute(0, 2, 1)

        graph_data = []
        for b in range(B):
            nodes = node_features[b]
            num_nodes = nodes.shape[0]

            # Enhanced connectivity: grid + k-nearest neighbors
            edge_indices = []
            edge_weights = []

            # Grid connectivity (8-connectivity)
            for i in range(H):
                for j in range(W):
                    node_idx = i * W + j
                    # 8-connectivity
                    for di in [-1, 0, 1]:
                        for dj in [-1, 0, 1]:
                            if di == 0 and dj == 0:
                                continue
                            ni, nj = i + di, j + dj
                            if 0 <= ni < H and 0 <= nj < W:
                                neighbor_idx = ni * W + nj
                                edge_indices.append([node_idx, neighbor_idx])

                                # Compute edge weight
                                node_pair = torch.cat([nodes[node_idx], nodes[neighbor_idx]])
                                weight = self.edge_proj(node_pair)
                                edge_weights.append(weight)

            if edge_indices:
                edge_index = torch.tensor(edge_indices, dtype=torch.long).t().to(nodes.device)
                edge_weight = torch.stack(edge_weights).squeeze().to(nodes.device)
            else:
                edge_index = torch.zeros((2, 0), dtype=torch.long, device=nodes.device)
                edge_weight = torch.zeros(0, device=nodes.device)

            graph_data.append({
                'x': nodes,
                'edge_index': edge_index,
                'edge_weight': edge_weight
            })

        return graph_data

class EnhancedGraphEncoder(nn.Module):
    """Enhanced Graph Neural Network with Custom GAT"""

    def __init__(self, node_dim=128, hidden_dim=256, num_layers=3, heads=4):
        super(EnhancedGraphEncoder, self).__init__()

        self.convs = nn.ModuleList()
        self.batch_norms = nn.ModuleList()

        # First GAT layer
        self.convs.append(CustomGraphAttention(node_dim, hidden_dim, heads=heads))
        self.batch_norms.append(nn.BatchNorm1d(hidden_dim))

        # Additional layers
        for _ in range(num_layers - 1):
            self.convs.append(CustomGraphAttention(hidden_dim, hidden_dim, heads=heads))
            self.batch_norms.append(nn.BatchNorm1d(hidden_dim))

        self.dropout = nn.Dropout(0.1)

    def forward(self, graph_list):
        if not graph_list:
            return None

        try:
            # Extract data from graph list
            x_list = [g['x'] for g in graph_list]
            edge_index_list = [g['edge_index'] for g in graph_list]
            edge_weight_list = [g.get('edge_weight', None) for g in graph_list]

            # Create batch
            batch = GraphBatch(x_list, edge_index_list, edge_weight_list)
            x, edge_index, batch_idx = batch.x, batch.edge_index, batch.batch

            for i, (conv, bn) in enumerate(zip(self.convs, self.batch_norms)):
                x = conv(x, edge_index)
                x = bn(x)
                x = F.relu(x)
                if i < len(self.convs) - 1:
                    x = self.dropout(x)

            # Global pooling with custom function
            graph_embeddings = global_mean_pool_custom(x, batch_idx)

            return graph_embeddings

        except Exception as e:
            print(f"Graph encoding error: {e}")
            return None

# ============================================================================
# SECTION 7: ENHANCED MAIN MODEL
# ============================================================================

class EnhancedFGANAD(nn.Module):
    """Enhanced FGAN-AD with better architecture"""

    def __init__(self, fpn_dim=256, node_dim=128, graph_hidden_dim=256):
        super(EnhancedFGANAD, self).__init__()

        # Enhanced Feature Pyramid Network
        self.fpn = EnhancedFPN()

        # Graph components for each FPN level
        self.graph_constructors = nn.ModuleList([
            EnhancedGraphConstructor(fpn_dim, node_dim) for _ in range(6)
        ])

        self.graph_encoders = nn.ModuleList([
            EnhancedGraphEncoder(node_dim, graph_hidden_dim) for _ in range(6)
        ])

        # Enhanced fusion layers
        self.fusion_layers = nn.ModuleList([
            nn.Sequential(
                nn.Linear(graph_hidden_dim, graph_hidden_dim // 2),
                nn.BatchNorm1d(graph_hidden_dim // 2),
                nn.ReLU(),
                nn.Dropout(0.1),
                nn.Linear(graph_hidden_dim // 2, graph_hidden_dim // 4),
                nn.BatchNorm1d(graph_hidden_dim // 4),
                nn.ReLU()
            ) for _ in range(6)
        ])

        # Multi-scale fusion
        fusion_dim = (graph_hidden_dim // 4) * 6
        self.scale_attention = nn.Sequential(
            nn.Linear(fusion_dim, fusion_dim // 2),
            nn.ReLU(),
            nn.Linear(fusion_dim // 2, 6),
            nn.Softmax(dim=1)
        )

        # Enhanced classification head
        self.classifier = nn.Sequential(
            nn.Linear(fusion_dim, fusion_dim // 2),
            nn.BatchNorm1d(fusion_dim // 2),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(fusion_dim // 2, fusion_dim // 4),
            nn.BatchNorm1d(fusion_dim // 4),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(fusion_dim // 4, fusion_dim // 8),
            nn.ReLU(),
            nn.Dropout(0.1),
            nn.Linear(fusion_dim // 8, 1),
            nn.Sigmoid()
        )

        # Enhanced reconstruction head
        self.reconstruction_head = nn.Sequential(
            nn.Linear(fusion_dim, 1024),
            nn.BatchNorm1d(1024),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(1024, 2048),
            nn.BatchNorm1d(2048),
            nn.ReLU(),
            nn.Dropout(0.1),
            nn.Linear(2048, 3 * 64 * 64),
            nn.Tanh()
        )

    def forward(self, x):
        batch_size = x.shape[0]

        # Extract multi-scale features
        fpn_features = self.fpn(x)

        # Process each level with graphs
        graph_embeddings = []
        valid_embeddings = []

        for i, (feature_map, constructor, encoder, fusion) in enumerate(
            zip(fpn_features, self.graph_constructors, self.graph_encoders, self.fusion_layers)
        ):
            try:
                graphs = constructor(feature_map)
                graph_embed = encoder(graphs)

                if graph_embed is not None and graph_embed.shape[0] == batch_size:
                    graph_embed = fusion(graph_embed)
                    graph_embeddings.append(graph_embed)
                    valid_embeddings.append(i)
                else:
                    # Fallback
                    graph_embeddings.append(torch.zeros(
                        batch_size, self.fusion_layers[0][-2].out_features,
                        device=x.device
                    ))
            except Exception as e:
                print(f"Warning: Error in graph processing at level {i}: {e}")
                graph_embeddings.append(torch.zeros(
                    batch_size, self.fusion_layers[0][-2].out_features,
                    device=x.device
                ))

        # Ensure we have all embeddings
        while len(graph_embeddings) < 6:
            graph_embeddings.append(torch.zeros_like(graph_embeddings[0]))

        # Multi-scale fusion with attention
        fused_features = torch.cat(graph_embeddings, dim=1)

        # Scale attention
        scale_weights = self.scale_attention(fused_features)
        weighted_features = []
        for i, (embed, weight) in enumerate(zip(graph_embeddings, scale_weights.chunk(6, dim=1))):
            weighted_features.append(embed * weight)

        final_features = torch.cat(weighted_features, dim=1)

        # Get outputs
        anomaly_scores = self.classifier(final_features)
        reconstructions = self.reconstruction_head(final_features)
        reconstructions = reconstructions.view(batch_size, 3, 64, 64)

        return anomaly_scores, reconstructions

# ============================================================================
# SECTION 8: ENHANCED LOSS FUNCTION
# ============================================================================

class EnhancedFGANADLoss(nn.Module):
    """Enhanced loss with multiple components"""

    def __init__(self, recon_weight=1.0, class_weight=1.0, contrast_weight=0.5):
        super(EnhancedFGANADLoss, self).__init__()
        self.recon_weight = recon_weight
        self.class_weight = class_weight
        self.contrast_weight = contrast_weight

        self.bce_loss = nn.BCELoss()
        self.mse_loss = nn.MSELoss()
        self.focal_loss_alpha = 0.25
        self.focal_loss_gamma = 2.0

    def focal_loss(self, pred, target):
        """Focal loss for class imbalance"""
        bce_loss = F.binary_cross_entropy(pred.squeeze(), target, reduction='none')
        pt = torch.exp(-bce_loss)
        focal_loss = self.focal_loss_alpha * (1-pt)**self.focal_loss_gamma * bce_loss
        return focal_loss.mean()

    def contrastive_loss(self, features, labels):
        """Contrastive loss to separate normal/anomaly features"""
        normal_features = features[labels == 0]
        anomaly_features = features[labels == 1]

        if len(normal_features) == 0 or len(anomaly_features) == 0:
            return torch.tensor(0.0, device=features.device)

        # Compute distances
        normal_center = normal_features.mean(dim=0)
        anomaly_center = anomaly_features.mean(dim=0)

        # Encourage separation
        center_distance = F.mse_loss(normal_center, anomaly_center)
        separation_loss = torch.exp(-center_distance)

        return separation_loss

    def forward(self, outputs, targets, images):
        anomaly_scores, reconstructions = outputs

        # Resize images for reconstruction loss
        images_resized = F.interpolate(images, size=(64, 64), mode='bilinear', align_corners=False)

        # Enhanced reconstruction loss with perceptual component
        recon_loss = self.mse_loss(reconstructions, images_resized)

        # L1 loss for sparsity
        l1_loss = F.l1_loss(reconstructions, images_resized)
        recon_loss = recon_loss + 0.1 * l1_loss

        # Classification loss
        if targets is not None and len(torch.unique(targets)) > 1:
            class_loss = self.focal_loss(anomaly_scores, targets)
        else:
            class_loss = torch.tensor(0.0, device=images.device)

        # Contrastive loss for feature separation
        if targets is not None and len(torch.unique(targets)) > 1:
            # Use reconstruction features for contrastive learning
            recon_features = reconstructions.view(reconstructions.shape[0], -1)
            contrast_loss = self.contrastive_loss(recon_features, targets)
        else:
            contrast_loss = torch.tensor(0.0, device=images.device)

        # Anomaly score regularization
        normal_mask = (targets == 0) if targets is not None else torch.ones_like(anomaly_scores.squeeze()).bool()
        if torch.any(normal_mask):
            # Encourage normal samples to have low scores
            normal_penalty = torch.mean(anomaly_scores[normal_mask])
        else:
            normal_penalty = torch.tensor(0.0, device=images.device)

        # Anomaly samples should have high scores
        anomaly_mask = (targets == 1) if targets is not None else torch.zeros_like(anomaly_scores.squeeze()).bool()
        if torch.any(anomaly_mask):
            anomaly_penalty = torch.mean(1.0 - anomaly_scores[anomaly_mask])
        else:
            anomaly_penalty = torch.tensor(0.0, device=images.device)

        # Combined loss
        total_loss = (self.recon_weight * recon_loss +
                     self.class_weight * class_loss +
                     self.contrast_weight * contrast_loss +
                     0.1 * normal_penalty +
                     0.1 * anomaly_penalty)

        return total_loss, {
            'total_loss': total_loss,
            'recon_loss': recon_loss,
            'class_loss': class_loss,
            'contrast_loss': contrast_loss,
            'normal_penalty': normal_penalty,
            'anomaly_penalty': anomaly_penalty
        }

# ============================================================================
# SECTION 9: ENHANCED TRAINING FUNCTIONS
# ============================================================================

class EnhancedTextureADDataset(Dataset):
    """Enhanced dataset class"""

    def __init__(self, images, labels=None, masks=None, transform=None):
        self.images = images
        self.labels = labels if labels is not None else [0] * len(images)
        self.masks = masks
        self.transform = transform

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        image = self.images[idx]
        label = self.labels[idx]

        if image.dtype != np.uint8:
            image = (image * 255).astype(np.uint8)

        if self.transform:
            transformed = self.transform(image=image)
            image = transformed['image']

        if not isinstance(image, torch.Tensor):
            image = torch.from_numpy(image).permute(2, 0, 1).float() / 255.0

        return {
            'image': image,
            'label': torch.tensor(label, dtype=torch.float32),
            'idx': idx
        }

def train_epoch_enhanced(model, dataloader, optimizer, criterion, device, scaler=None):
    """Enhanced training with mixed precision"""
    model.train()
    total_loss = 0
    loss_components = {
        'recon_loss': 0, 'class_loss': 0, 'contrast_loss': 0,
        'normal_penalty': 0, 'anomaly_penalty': 0
    }

    for batch in tqdm(dataloader, desc='Training'):
        images = batch['image'].to(device)
        labels = batch['label'].to(device)

        optimizer.zero_grad()

        if scaler is not None:
            with torch.cuda.amp.autocast():
                anomaly_scores, reconstructions = model(images)
                loss, loss_dict = criterion((anomaly_scores, reconstructions), labels, images)

            scaler.scale(loss).backward()
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            scaler.step(optimizer)
            scaler.update()
        else:
            anomaly_scores, reconstructions = model(images)
            loss, loss_dict = criterion((anomaly_scores, reconstructions), labels, images)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            optimizer.step()

        total_loss += loss.item()
        for key in loss_components:
            if key in loss_dict:
                loss_components[key] += loss_dict[key].item()

    avg_loss = total_loss / len(dataloader)
    for key in loss_components:
        loss_components[key] /= len(dataloader)

    return avg_loss, loss_components

def evaluate_model_enhanced(model, dataloader, criterion, device):
    """Enhanced evaluation with detailed metrics"""
    model.eval()
    total_loss = 0
    all_scores = []
    all_labels = []

    with torch.no_grad():
        for batch in tqdm(dataloader, desc='Evaluating'):
            images = batch['image'].to(device)
            labels = batch['label'].to(device)

            anomaly_scores, reconstructions = model(images)
            loss, _ = criterion((anomaly_scores, reconstructions), labels, images)

            total_loss += loss.item()
            all_scores.extend(anomaly_scores.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    avg_loss = total_loss / len(dataloader)
    all_scores = np.array(all_scores).flatten()
    all_labels = np.array(all_labels)

    # Compute metrics
    metrics = {}
    if len(np.unique(all_labels)) > 1:
        # AUC-ROC
        metrics['auc_roc'] = roc_auc_score(all_labels, all_scores)

        # AUC-PR
        precision, recall, _ = precision_recall_curve(all_labels, all_scores)
        metrics['auc_pr'] = auc(recall, precision)

        # Best F1 score
        thresholds = np.linspace(0, 1, 100)
        f1_scores = []
        for thresh in thresholds:
            pred_labels = (all_scores > thresh).astype(int)
            if len(np.unique(pred_labels)) > 1:
                f1 = f1_score(all_labels, pred_labels)
                f1_scores.append(f1)
            else:
                f1_scores.append(0)

        metrics['best_f1'] = max(f1_scores)
        metrics['best_threshold'] = thresholds[np.argmax(f1_scores)]
    else:
        metrics['auc_roc'] = 0.0
        metrics['auc_pr'] = 0.0
        metrics['best_f1'] = 0.0
        metrics['best_threshold'] = 0.5

    return avg_loss, metrics, all_scores, all_labels

def visualize_enhanced_results(model, test_loader, device, num_samples=8):
    """Enhanced visualization with better analysis"""
    model.eval()
    fig, axes = plt.subplots(4, num_samples, figsize=(20, 12))

    with torch.no_grad():
        # Get a mix of normal and anomaly samples
        normal_batch = None
        anomaly_batch = None

        for batch in test_loader:
            labels = batch['label']
            if normal_batch is None and torch.any(labels == 0):
                normal_indices = torch.where(labels == 0)[0][:num_samples//2]
                normal_batch = {
                    'image': batch['image'][normal_indices],
                    'label': batch['label'][normal_indices]
                }

            if anomaly_batch is None and torch.any(labels == 1):
                anomaly_indices = torch.where(labels == 1)[0][:num_samples//2]
                anomaly_batch = {
                    'image': batch['image'][anomaly_indices],
                    'label': batch['label'][anomaly_indices]
                }

            if normal_batch is not None and anomaly_batch is not None:
                break

        # Combine batches
        if normal_batch is not None and anomaly_batch is not None:
            images = torch.cat([normal_batch['image'], anomaly_batch['image']])
            labels = torch.cat([normal_batch['label'], anomaly_batch['label']])
        elif normal_batch is not None:
            images = normal_batch['image'][:num_samples]
            labels = normal_batch['label'][:num_samples]
        else:
            batch = next(iter(test_loader))
            images = batch['image'][:num_samples]
            labels = batch['label'][:num_samples]

        images = images.to(device)
        anomaly_scores, reconstructions = model(images)

        # Denormalize for visualization
        mean = torch.tensor([0.485, 0.456, 0.406]).view(1, 3, 1, 1).to(device)
        std = torch.tensor([0.229, 0.224, 0.225]).view(1, 3, 1, 1).to(device)

        images_vis = images * std + mean
        images_vis = torch.clamp(images_vis, 0, 1)

        reconstructions_vis = reconstructions * std + mean
        reconstructions_vis = torch.clamp(reconstructions_vis, 0, 1)

        for i in range(min(num_samples, len(images))):
            # Original
            axes[0, i].imshow(images_vis[i].cpu().permute(1, 2, 0))
            status = "Normal" if labels[i] == 0 else "Anomaly"
            axes[0, i].set_title(f'Original ({status})')
            axes[0, i].axis('off')

            # Reconstruction
            axes[1, i].imshow(reconstructions_vis[i].cpu().permute(1, 2, 0))
            axes[1, i].set_title('Reconstruction')
            axes[1, i].axis('off')

            # Error map
            original_resized = F.interpolate(
                images_vis[i:i+1], size=(64, 64), mode='bilinear', align_corners=False
            )[0]
            error = torch.mean((original_resized - reconstructions_vis[i])**2, dim=0)

            im = axes[2, i].imshow(error.cpu(), cmap='hot')
            axes[2, i].set_title(f'Error Map')
            axes[2, i].axis('off')

            # Anomaly score visualization
            score = anomaly_scores[i].item()
            color = 'red' if score > 0.5 else 'green'
            axes[3, i].bar([0], [score], color=color, alpha=0.7)
            axes[3, i].set_ylim(0, 1)
            axes[3, i].set_title(f'Score: {score:.3f}')
            axes[3, i].set_xticks([])

    plt.tight_layout()
    plt.show()

# ============================================================================
# SECTION 10: MAIN EXECUTION
# ============================================================================

def main_enhanced_training():
    """Main function for enhanced training"""

    print("🏗️ Creating enhanced dataset...")
    (train_images, train_labels), (test_images, test_labels) = create_enhanced_dataset()

    # Create enhanced transforms
    train_transform, val_transform = get_enhanced_transforms()

    # Create datasets
    train_dataset = EnhancedTextureADDataset(train_images, train_labels, transform=train_transform)
    test_dataset = EnhancedTextureADDataset(test_images, test_labels, transform=val_transform)

    # Enhanced data loaders with better sampling
    batch_size = 16  # Increased batch size
    train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True,
                             num_workers=4, pin_memory=True, drop_last=True)
    test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False,
                            num_workers=4, pin_memory=True)

    print(f"📦 Enhanced data loaders: {len(train_loader)} train, {len(test_loader)} test batches")

    # Initialize enhanced model
    print("🧠 Initializing Enhanced FGAN-AD model...")
    model = EnhancedFGANAD().to(device)

    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"🔢 Total parameters: {total_params:,}")
    print(f"🔢 Trainable parameters: {trainable_params:,}")

    # Enhanced training setup
    optimizer = optim.AdamW(model.parameters(), lr=2e-4, weight_decay=1e-4, betas=(0.9, 0.999))

    # Cosine annealing with warm restarts
    scheduler = optim.lr_scheduler.CosineAnnealingWarmRestarts(
        optimizer, T_0=10, T_mult=2, eta_min=1e-6
    )

    criterion = EnhancedFGANADLoss(recon_weight=1.0, class_weight=2.0, contrast_weight=0.5)

    # Mixed precision training
    scaler = torch.cuda.amp.GradScaler() if device.type == 'cuda' else None

    # Training with enhanced parameters
    num_epochs = 50  # Increased epochs
    best_auc = 0.0
    patience = 15
    patience_counter = 0

    train_losses = []
    val_aucs = []
    val_f1s = []

    print(f"🚂 Starting enhanced training for {num_epochs} epochs...")
    print("="*80)

    for epoch in range(num_epochs):
        # Training
        train_loss, loss_components = train_epoch_enhanced(
            model, train_loader, optimizer, criterion, device, scaler
        )

        # Evaluation
        val_loss, metrics, val_scores, val_labels = evaluate_model_enhanced(
            model, test_loader, criterion, device
        )

        # Scheduler step
        scheduler.step()

        # Store metrics
        train_losses.append(train_loss)
        val_aucs.append(metrics['auc_roc'])
        val_f1s.append(metrics['best_f1'])

        # Print detailed results
        print(f"Epoch {epoch+1:2d}/{num_epochs}")
        print(f"  Train Loss: {train_loss:.4f}")
        print(f"  Val Loss: {val_loss:.4f}")
        print(f"  Val AUC-ROC: {metrics['auc_roc']:.4f}")
        print(f"  Val AUC-PR: {metrics['auc_pr']:.4f}")
        print(f"  Val F1: {metrics['best_f1']:.4f}")
        print(f"  Best Threshold: {metrics['best_threshold']:.3f}")
        print(f"  LR: {optimizer.param_groups[0]['lr']:.2e}")

        # Early stopping and model saving
        if metrics['auc_roc'] > best_auc:
            best_auc = metrics['auc_roc']
            patience_counter = 0
            torch.save({
                'epoch': epoch,
                'model_state_dict': model.state_dict(),
                'optimizer_state_dict': optimizer.state_dict(),
                'best_auc': best_auc,
                'metrics': metrics
            }, 'best_enhanced_fgan_ad_model.pth')
            print(f"🏆 New best model saved! AUC: {best_auc:.4f}")
        else:
            patience_counter += 1

        if patience_counter >= patience:
            print(f"⏰ Early stopping triggered after {patience} epochs without improvement")
            break

        # Visualize every 15 epochs
        if (epoch + 1) % 15 == 0:
            print("\n📊 Visualizing enhanced results...")
            visualize_enhanced_results(model, test_loader, device)

            # Plot enhanced training curves
            fig, axes = plt.subplots(2, 2, figsize=(15, 10))

            # Training loss
            axes[0, 0].plot(train_losses, label='Train Loss', color='blue')
            axes[0, 0].set_title('Training Loss')
            axes[0, 0].set_xlabel('Epoch')
            axes[0, 0].set_ylabel('Loss')
            axes[0, 0].legend()
            axes[0, 0].grid(True)

            # Validation AUC
            axes[0, 1].plot(val_aucs, label='Validation AUC-ROC', color='orange')
            axes[0, 1].axhline(y=0.98, color='red', linestyle='--', label='Target 98%')
            axes[0, 1].set_title('Validation AUC-ROC')
            axes[0, 1].set_xlabel('Epoch')
            axes[0, 1].set_ylabel('AUC-ROC')
            axes[0, 1].legend()
            axes[0, 1].grid(True)

            # F1 Score
            axes[1, 0].plot(val_f1s, label='Validation F1', color='green')
            axes[1, 0].set_title('Validation F1 Score')
            axes[1, 0].set_xlabel('Epoch')
            axes[1, 0].set_ylabel('F1 Score')
            axes[1, 0].legend()
            axes[1, 0].grid(True)

            # Current progress
            current_progress = best_auc / 0.98 * 100
            axes[1, 1].bar(['Progress'], [current_progress], color='purple', alpha=0.7)
            axes[1, 1].axhline(y=100, color='red', linestyle='--', label='98% Target')
            axes[1, 1].set_title('Progress to 98% AUC Target')
            axes[1, 1].set_ylabel('Progress (%)')
            axes[1, 1].set_ylim(0, 110)
            axes[1, 1].legend()
            axes[1, 1].grid(True)

            plt.tight_layout()
            plt.show()

        print("-" * 40)

    print(f"\n🎉 Enhanced training completed!")
    print(f"🏆 Best AUC achieved: {best_auc:.4f}")

    if best_auc >= 0.98:
        print("🎯 TARGET ACHIEVED: 98%+ AUC!")
        print("🌟 Congratulations! Your model has reached the target performance!")
    else:
        progress = best_auc / 0.98 * 100
        print(f"📈 Progress: {progress:.1f}% towards 98% target")
        print(f"💡 You're {100-progress:.1f}% away from the target!")

    return model, best_auc, metrics

# Execute enhanced training
if __name__ == "__main__":
    print("🚀 Starting Enhanced FGAN-AD Training...")
    set_seed(42)
    model, best_auc, final_metrics = main_enhanced_training()

    print("\n" + "="*60)
    print("🎯 FINAL RESULTS:")
    print(f"🏆 Best AUC-ROC: {best_auc:.4f}")
    print(f"🎯 Target: 0.9800")
    print(f"📊 Progress: {best_auc/0.98*100:.1f}%")

    if best_auc >= 0.98:
        print("✅ SUCCESS: 98% TARGET ACHIEVED! 🎉")
    else:
        print(f"⚠️  Almost there! Need {0.98-best_auc:.4f} more AUC points")
    print("="*60)
```
